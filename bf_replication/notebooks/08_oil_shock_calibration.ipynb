# Notebook 08 - Historical oil shock calibration

Port of `Oil_Shock.m`. Uses the **1971** IO table with different elasticities:
- ε = 0.02 (near-Leontief between labor and intermediates)
- θ = 0.25
- σ = 0.25

The very low ε creates strong amplification because the economy cannot
substitute away from oil.

```julia
using LinearAlgebra, Statistics, Printf, DelimitedFiles

const NB_DIR = @__DIR__
const PKG = joinpath(NB_DIR, "..")
const DATA_DIR = joinpath(PKG, "..", "Replication Files", "GDP Simulatin -- 88 Sector")
const RESULTS_DIR = joinpath(PKG, "data", "results")
mkpath(RESULTS_DIR)
push!(LOAD_PATH, joinpath(PKG, "src"))
include(joinpath(PKG, "src", "BFReplication.jl"))
using .BFReplication
using .BFReplication.DataLoader

println("Module loaded OK")
```

```julia
# Load 1971 IO data (Oil_Shock.m uses year=1971)
data71 = load_bf_data(joinpath(DATA_DIR, "BFdata.csv"); year=1971)
println("1971 IO data: N=$(data71.N) sectors")

# Also load 1980 for comparison
data80 = load_bf_data(joinpath(DATA_DIR, "BFdata.csv"); year=1980)
println("1980 IO data: N=$(data80.N) sectors")
```

```julia
# Oil shock: MATLAB uses A[7] = 0.9 (10% negative shock)
# Parameters: epsilon = 0.02, theta = 0.25, sigma = 0.25
println("=== Oil shock (1971, MATLAB calibration) ===")
r = run_oil_shock(data71; A_shock_val=0.9, ε=0.02, θ=0.25, σ=0.25)
println("  Shock: A[7] = 0.9 (10% negative)")
println("  GDP              = $(round(r.gdp, digits=6))")
println("  Δlog GDP         = $(round(r.log_gdp_change, digits=6))  ($(round(r.log_gdp_change*100, digits=2))%)")
println("  Hulten           = $(round(r.hulten, digits=6))")
println("  Amplification    = $(round(r.amplification, digits=3))x")
```

```julia
# Compare with 30% shock (like the paper's oil shock figure)
println("=== Oil shock (1971, A=0.7, paper-style) ===")
r2 = run_oil_shock(data71; A_shock_val=0.7, ε=0.02, θ=0.25, σ=0.25)
println("  Δlog GDP         = $(round(r2.log_gdp_change, digits=6))")
println("  Amplification    = $(round(r2.amplification, digits=3))x")
println()

# Compare with 1980 paper calibration
println("=== Comparison: same shock (A=0.7), different calibrations ===")
r3 = run_oil_shock(data80; A_shock_val=0.7, ε=0.5, θ=0.001, σ=0.9)
println("  1980 paper calib (ε=0.5 θ=0.001 σ=0.9):")
println("    Δlog GDP = $(round(r3.log_gdp_change, digits=6)), amp = $(round(r3.amplification, digits=3))x")
println("  1971 oil calib (ε=0.02 θ=0.25 σ=0.25):")
println("    Δlog GDP = $(round(r2.log_gdp_change, digits=6)), amp = $(round(r2.amplification, digits=3))x")
```
